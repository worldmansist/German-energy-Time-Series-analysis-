```python
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Загружаем данные
iris = load_iris()

# Преобразуем в DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Добавим колонку target (классы)
df['target'] = iris.target

# Смотрим на первые строки
df.head()
```

```python
import seaborn as sns
# Картинки в ноутбуке
%matplotlib inline

# 1. Общая информация
print("\n Shape:", df.shape)
print("\n Колонки:", df.columns.tolist())
print("\n Типы данных:")
print(df.dtypes)
print("\n Пропуски:")
print(df.isnull().sum())
```

```python
# 2. Статистика
print("\n Статистика:")
print(df.describe())

# 3. Распределение классов
print("\n Распределение классов:")
print(df['target'].value_counts())
```

```python
# 4. Pairplot (взаимосвязь признаков и классов)
sns.pairplot(df, hue="target", diag_kind="kde", palette="tab10")
plt.suptitle("🔗 Парные графики признаков", y=1.02)
plt.show()

# 5. Heatmap корреляций
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("🔥 Корреляции между признаками")
plt.show()

# 6. Boxplot по классам
plt.figure(figsize=(12, 6))
df_melt = pd.melt(df, id_vars="target", var_name="feature", value_name="value")
sns.boxplot(x="feature", y="value", hue="target", data=df_melt, palette="tab10")
plt.title("📊 Распределение признаков по классам")
plt.xticks(rotation=45)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Разделяем данные на train/test
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, random_state=42
)

# Обучаем модель
model = RandomForestClassifier(n_estimators=100, max_depth=3)
model.fit(X_train, y_train)

# Оцениваем
preds = model.predict(X_test)
acc = accuracy_score(y_test, preds)

print(f"Accuracy: {acc}")
```

```python
import mlflow
import mlflow.sklearn

# Запускаем логирование
with mlflow.start_run():

    # Параметры
    n_estimators = 100
    max_depth = 3

    # Обучаем модель
    model = RandomForestClassifier(n_estimators=n_estimators, max_depth=max_depth)
    model.fit(X_train, y_train)

    # Оценка
    preds = model.predict(X_test)
    acc = accuracy_score(y_test, preds)

    # Логируем параметры
    mlflow.log_param("n_estimators", n_estimators)
    mlflow.log_param("max_depth", max_depth)

    # Логируем метрику
    mlflow.log_metric("accuracy", acc)

    # Логируем модель
    mlflow.sklearn.log_model(model, "model")

    print(f"Accuracy: {acc}")
```
