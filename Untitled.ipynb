```python
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
from pmdarima import auto_arima
from statsmodels.tsa.statespace.sarimax import SARIMAX
```

```python
# === 1. Прямой путь к файлу относительно проекта ===
data_path = "C:/Users/user/streamlit-crash-cours/data/opsd_germany_daily.csv"

# === 2. Загружаем данные ===
df = pd.read_csv(data_path)

print('Shape', df.shape)
print(df.head())
```

```python
# === 3. Преобразуем даты ===
df["Date"] = pd.to_datetime(df["Date"])
df.set_index("Date", inplace=True)
```

```python
# Для примера возьмем колонку "Consumption"
series = df["Consumption"].dropna()
series_plot = series.iloc[:90]
# === 4. Визуализация ===
plt.figure(figsize=(12, 5))
series_plot.plot(title="Germany Daily Electricity Consumption")
plt.xlabel("Date")
plt.ylabel("Consumption (MWh)")
plt.show()
```

```python
# === 5. Train / Test Split ===
train = series.iloc[:-7]
test = series.iloc[-7:]
```

```python
train
```

```python
# === 4. Автоподбор параметров SARIMAX ===
stepwise_model = auto_arima(
    train,
    start_p=0, max_p=3,
    start_q=0, max_q=3,
    start_P=0, max_P=2,
    start_Q=0, max_Q=2,
    d=None,
    D=None,
    seasonal=True,
    m=7,                # годовая сезонность
    stepwise=True,
    trace=True,
    error_action="ignore",
    suppress_warnings=True,
    n_jobs=-1
)

print("Best parameters:", stepwise_model.order, stepwise_model.seasonal_order)

# === 5. Обучение финальной модели ===
model = SARIMAX(train,
                order=stepwise_model.order,
                seasonal_order=stepwise_model.seasonal_order)
model_fit = model.fit(disp=False)
```

```python
import joblib

# Сохранение
joblib.dump(model_fit, "sarimax_model.joblib")

# Загрузка
loaded_model = joblib.load("sarimax_model.joblib")

forecast = loaded_model.forecast(steps=7)
```

```python
forecast_plot = forecast.iloc[-7:]
train_plot = train.iloc[-7:]
test_plot = test.iloc[::]

# === 8. Визуализация прогноза ===
plt.figure(figsize=(12, 5))
plt.plot(train_plot.index, train_plot, label="Train")
plt.plot(test_plot.index, test_plot, label="Test", color="green")
plt.plot(test_plot.index, forecast_plot, label="Forecast", color="red")
plt.title("SARIMAX Forecast vs Actual")
plt.legend()
plt.show()
```
