```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Downloading Load
df_Load = dfd['Load (incl. self-consumption)']
df_Load.name = "Load"
df_Load_2023 = df_Load.loc["2022-06-01":]
dfDP = df_Load_2023.copy().to_frame()

# Scaling Load
scaler = MinMaxScaler()
dfDP['Load'] = scaler.fit_transform(dfDP[['Load']])

#
dfDP.index = pd.to_datetime(dfDP.index)

# Add external features
dfDP['hour'] = dfDP.index.hour
dfDP['dayofweek'] = dfDP.index.dayofweek
dfDP['month'] = dfDP.index.month
dfDP['is_weekend'] = (dfDP.index.dayofweek >= 5).astype(int)

# Normalize external features
scaler_ext = MinMaxScaler()
external_features = scaler_ext.fit_transform(dfDP[['hour', 'dayofweek', 'month', 'is_weekend']])
dfDP[['hour', 'dayofweek', 'month', 'is_weekend']] = external_features

# List of features
feature_cols = ['Load', 'hour', 'dayofweek', 'month', 'is_weekend']
ylist = dfDP['Load'].tolist()
X_features = dfDP[feature_cols].values

# parameters
n_future = 96
n_past = 672
total_period = n_past + n_future


X_new = []
y_new = []
idx_end = len(dfDP)
idx_start = idx_end - total_period

while idx_start > 0:

    x_window = X_features[idx_start:idx_start + n_past, :]  # shape = (n_past, n_features)
    y_window = ylist[idx_start + n_past:idx_start + total_period]

    X_new.append(x_window)
    y_new.append(y_window)

    idx_start -= 1

X_new = np.array(X_new)  # shape = (samples, n_past, n_features)
y_new = np.array(y_new)  # shape = (samples, n_future)

# Train/test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_new, y_new, test_size=0.33, random_state=42)


X_train_rs = X_train
X_test_rs = X_test


n_samples = X_train_rs.shape[0]
n_timesteps = X_train_rs.shape[1]
n_features = X_train_rs.shape[2]
n_steps = y_train.shape[1]

n_timesteps = X_train_rs.shape[1]
n_features = X_train_rs.shape[2]
batch_size = 64
```

```python
import keras
import random
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM, Dropout, SimpleRNN
from tensorflow.keras.callbacks import EarlyStopping


random.seed(42)

simpleRNN_model = Sequential([
  SimpleRNN(64, activation='tanh', input_shape=(n_timesteps, n_features), return_sequences=True),
  BatchNormalization(),
  Dropout(0.3),
  SimpleRNN(64, activation='tanh', return_sequences=True),
  BatchNormalization(),
  Dropout(0.2),
  SimpleRNN(64, activation='tanh'),
  Dense(y_train.shape[1]),
])

simpleRNN_model.compile(
  optimizer=keras.optimizers.Adam(learning_rate=0.001),
  loss='mean_absolute_error',
  metrics=['mean_absolute_error'],
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=5,
    min_delta=0.001,
    restore_best_weights=True
)

smod_history = simpleRNN_model.fit(
    X_train_rs, y_train,
    validation_split=0.2,
    epochs=15,
    batch_size=batch_size,
    shuffle=True,
    callbacks=[early_stopping]
)

preds = simpleRNN_model.predict(X_test_rs)

print(f"R2 score: {r2_score(y_test, preds):.4f}")

plt.plot(smod_history.history['loss'])
plt.plot(smod_history.history['val_loss'])
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')
plt.show()
```

```python
import random
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM, Dropout, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow import keras
import matplotlib.pyplot as plt

random.seed(42)

simpleLSTM_model = Sequential([
    LSTM(128, activation='tanh', input_shape=(n_timesteps, n_features), return_sequences=True),
    BatchNormalization(),
    Dropout(0.3),
    LSTM(64, activation='tanh'),
    BatchNormalization(),
    Dropout(0.2),
    Dense(y_train.shape[1]),
])

simpleLSTM_model.summary()

simpleLSTM_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='mean_squared_error',
    metrics=['mean_absolute_error'],
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=2,
    min_delta=0.0005,
    restore_best_weights=True
)

smod_history = simpleLSTM_model.fit(
    X_train_rs, y_train,
    validation_split=0.2,
    epochs=15,
    batch_size=batch_size,
    shuffle=True,
    callbacks=[early_stopping]
)


preds = simpleLSTM_model.predict(X_test_rs)


plt.plot(smod_history.history['loss'])
plt.plot(smod_history.history['val_loss'])
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')
plt.show()
```
